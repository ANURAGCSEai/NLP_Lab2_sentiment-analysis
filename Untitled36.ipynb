## Sentiment Analysis using BERT

This section demonstrates how to perform sentiment analysis using a pre-trained BERT model from the Hugging Face `transformers` library.

```python
# Install necessary libraries
!pip install transformers torch pandas
```

```python
import torch
from transformers import pipeline

# Load a pre-trained sentiment analysis pipeline
# We'll use a model fine-tuned for sentiment analysis (e.g., 'distilbert-base-uncased-finetuned-sst-2-english')
sentiment_pipeline = pipeline("sentiment-analysis")

print("Sentiment analysis pipeline loaded successfully!")
```

### Example Usage

```python
# Example sentences for sentiment analysis
texts = [
    "I love this product, it's amazing!",
    "This movie was just okay, nothing special.",
    "I absolutely hate this service, it's terrible.",
    "The weather is beautiful today."
]

# Perform sentiment analysis
results = sentiment_pipeline(texts)

# Display the results
for text, result in zip(texts, results):
    print(f"Text: '{text}'")
    print(f"Sentiment: {result['label']}, Score: {result['score']:.4f}")
    print("-" * 30)
```
